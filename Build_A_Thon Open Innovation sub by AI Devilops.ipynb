## **Problem Statement:**

*Traditional dating platforms primarily rely on demographics, photos, location, and generic interests to match people. While these factors may indicate surface-level compatibility, they often fail to capture an important aspect of modern relationships: shared humor, internet culture, and personality expression.*

*In today's digital world, people connect through memes, fandoms, gaming communities, online trends, niche interests, and social media humor. Two individuals may have completely different backgrounds but instantly bond over the same meme format, anime reference, coding joke, or internet trend. However, existing matchmaking systems rarely understand these cultural signals.*

The challenge is to build an AI-powered matchmaking experience that goes beyond traditional profiles and identifies compatibility through users' interests, humor preferences, and online personality traits.

The system should:

* Analyze user-selected interests and mood boards.
* Infer personality traits and humor styles.
* Identify shared and complementary interests.
* Estimate compatibility between two users.
* Generate personalized meme-based pickup lines and icebreakers.
* Present matchmaking results in a fun, engaging, and shareable format.

The goal is to create a dating experience where people are matched not just by who they are, but by how they laugh, communicate, and engage with internet culture.

```python
# =========================
# IMPORTS
# =========================
import os
import sys
from typing import Any, List, Optional
from langchain_core.language_models.llms import BaseLLM as LangChainLLM
from langchain_core.output_parsers import StrOutputParser
from langchain_core.outputs import Generation, LLMResult
from langchain_core.prompts import PromptTemplate
from txtai.pipeline import LLM as TxtaiLLM

# =========================
# LOAD LOCAL LLM
# =========================
txtai_llm_instance = TxtaiLLM(
    "LiteLLMs/gemma-2b-it-GGUF/Q8_0/Q8_0-00001-of-00001.gguf"
)

# =========================
# LANGCHAIN WRAPPER
# =========================
class TxtaiLangChainLLM(LangChainLLM):

    @property
    def _llm_type(self) -> str:
        return "txtai"

    def _generate(
        self,
        prompts: List[str],
        stop: Optional[List[str]] = None,
        **kwargs: Any
    ) -> LLMResult:

        generations = []

        for prompt in prompts:
            response = txtai_llm_instance(prompt)
            generations.append(
                [Generation(text=response)]
            )

        return LLMResult(generations=generations)


llm = TxtaiLangChainLLM()

# =========================
# MOOD BOARD OPTIONS
# =========================
MOOD_BOARD_OPTIONS = {
    1: "Software Engineer",
    2: "Data Scientist",
    3: "AI Enthusiast",
    4: "Entrepreneur",
    5: "Designer",
    6: "Gaming",
    7: "Reading",
    8: "Gym/Fitness",
    9: "Traveling",
    10: "Photography",
    11: "Anime",
    12: "Movies",
    13: "Netflix",
    14: "Music",
    15: "Coffee",
    16: "Tea",
    17: "Pizza",
    18: "Street Food",
    19: "Chocolate",
    20: "Dark Humor",
    21: "Sarcasm",
    22: "Wholesome Memes",
    23: "Gen-Z Humor",
    24: "Nerd Humor"
}

# =========================
# MEME DATING PROMPT
# =========================
MEME_DATING_PROMPT = """
You are the Ultimate Meme Dating App AI Assistant.

Your mission is to determine whether two people are destined to become:
* Meme Soulmates
* Chaos Partners
* Co-op Teammates
* Main Character Duo
* Or a Complete Internet Disaster

Analyze both users using their interests, mood boards, and overall internet energy.

=================================================
MEME PERSONALITY ANALYSIS
=========================

Determine a Meme Personality Type for each user.

Use these archetypes when applicable:

Tech Interests
(AI, Coding, Programming, ML, Data Science, Startups)
→ The Caffeinated Tech Goblin

Gaming Interests
(Gaming, Esports, Streaming)
→ The Respawn Romantic

Fitness Interests
(Gym, Fitness, Sports, Workout)
→ The Pre-Workout Philosopher

Entertainment Interests
(Movies, Anime, Netflix, Series)
→ The Pop-Culture Archivist

Creative Interests
(Photography, Design, Art, Music)
→ The Creative Chaos Engineer

Books & Writing Interests
(Books, Reading, Poetry, Writing)
→ The Overthinking Storyteller

Travel & Adventure Interests
(Traveling, Hiking, Adventure)
→ The Wandering Memory Collector

Food & Beverage Interests
(Coffee, Tea, Pizza, Street Food, Chocolate)
→ The Comfort-Core Connoisseur

If multiple categories overlap:
Create a unique hybrid archetype.

For each user infer:
* Personality Traits
* Humor Style
* Communication Style
* Internet Behavior
* Meme Personality Type

=================================================
COMPATIBILITY SCORING
=====================

Generate a compatibility score between 0 and 100.

Scoring Rules:
Shared Interests > 4 -> Score between 85-100
Shared Interests between 2 and 4 -> Score between 70-85
Shared Interests = 1 -> Score between 55-70
No common interests but complementary personalities -> Score between 50-75
No common interests and conflicting personalities -> Score below 50

=================================================
IMPORTANT
=========
Return ONLY the requested output.
Do NOT explain your reasoning.
Do NOT repeat rules.
Do NOT mention scoring logic.
Do NOT add extra sections.

=================================================
OUTPUT FORMAT
=============
Follow this exact formatting. Replace names dynamically with the names provided in the input.

## {user_1_name} and {user_2_name}'s Meme Match Report

**MEME PERSONALITY:**

* User 1: <1-sentence description of User 1's vibe, traits, and archetype>
* User 2: <1-sentence description of User 2's vibe, traits, and archetype>

**COMPATIBILITY SCORE:** <0-100>%

**WHY THEY MATCH:**

<2-3 concise sentences explaining how their specific interests or personalities mesh together dynamically.>

**SHARED INTERESTS:**

<Bullet points of common interests, or if none, bullet points of complementary interest paths>

**MEME PICKUP LINE:**

"<A short, funny, copy-paste ready meme pickup line using their interests>"

**ICEBREAKER:**

"<A witty, natural opening message utilizing their interests with smooth emoji use>"

=================================================
INPUT
=====

User 1 Name:
{user_1_name}

User 1 Age:
{user_1_age}

User 1 Gender:
{user_1_gender}

User 1 Interests:
{user_1_interests}

---

User 2 Name:
{user_2_name}

User 2 Age:
{user_2_age}

User 2 Gender:
{user_2_gender}

User 2 Interests:
{user_2_interests}

"""

# =========================
# SELECT INTERESTS
# =========================
def select_interests():

    print("\n🎭 AVAILABLE INTERESTS")
    print("=" * 50)

    for key, value in MOOD_BOARD_OPTIONS.items():
        print(f"{key}. {value}")

    print("\nExample: 1,3,6,15,21")

    selected = input(
        "\nSelect interests: "
    ).strip()

    interests = []

    try:
        choices = [
            int(x.strip())
            for x in selected.split(",")
        ]

        for choice in choices:
            if choice in MOOD_BOARD_OPTIONS:
                interests.append(
                    MOOD_BOARD_OPTIONS[choice]
                )
    except:
        pass

    custom = input(
        "\n✨ Add custom interests (optional): "
    ).strip()

    if custom:
        custom_interests = [
            x.strip()
            for x in custom.split(",")
            if x.strip()
        ]
        interests.extend(custom_interests)

    return interests


# =========================
# USER INPUT
# =========================
def get_user_details(user_label):

    print(f"\n💕 Enter {user_label} Details")
    print("=" * 50)

    data = {}
    data["name"] = input(
        "👤 Name: "
    ).strip()

    while True:
        age = input(
            "🎂 Age: "
        ).strip()

        if age.isdigit():
            data["age"] = int(age)
            break

        print("❌ Enter valid age.")

    data["gender"] = input(
        "🚻 Gender: "
    ).strip()

    data["interests"] = select_interests()

    return data


# =========================
# RUN MATCHMAKING
# =========================
def run_matchmaking(user_1, user_2):

    prompt = PromptTemplate.from_template(
        MEME_DATING_PROMPT
    )

    chain = (
        prompt
        | llm
        | StrOutputParser()
    )

    result = chain.invoke({
        "user_1_name": user_1["name"],
        "user_1_age": user_1["age"],
        "user_1_gender": user_1["gender"],
        "user_1_interests": ", ".join(user_1["interests"]),

        "user_2_name": user_2["name"],
        "user_2_age": user_2["age"],
        "user_2_gender": user_2["gender"],
        "user_2_interests": ", ".join(user_2["interests"]),
    })

    return result

# =========================
# DISPLAY MATCH RESULT
# =========================
def display_match_result(user_1, user_2, report):

    # Clear terminal screen
    os.system("cls" if os.name == "nt" else "clear")

    print("\n")
    print("💘" + "=" * 66 + "💘")
    print("                    MEME MATCH FOUND")
    print("💘" + "=" * 66 + "💘")

    print(
        f"\n🎯 Match Candidate: "
        f"{user_1['name']} ❤️ {user_2['name']}"
    )

    print("\n" + "=" * 70 + "\n")

    print(report.strip())

    print("\n" + "=" * 70)
    print("🚀 Powered by LLMates AI")
    print("=" * 70)



# =========================
# MAIN
# =========================
def main():

    print("\n💘 MEME DATING APP 💘")
    print("=" * 60)

    user_1 = get_user_details("User 1")
    user_2 = get_user_details("User 2")

    print("\n⏳ Finding your meme soulmate...\n")

    report = run_matchmaking(
        user_1,
        user_2
    )

    display_match_result(
        user_1,
        user_2,
        report
    )
# =========================
# ENTRY POINT
# =========================
if __name__ == "__main__":
    main()
```

```python
# =========================
# IMPORTS
# =========================
from typing import Any, List, Optional
import gradio as gr
from langchain_core.language_models.llms import BaseLLM as LangChainLLM
from langchain_core.output_parsers import StrOutputParser
from langchain_core.outputs import Generation, LLMResult
from langchain_core.prompts import PromptTemplate
from txtai.pipeline import LLM as TxtaiLLM

# =========================
# LOAD LOCAL LLM
# =========================
txtai_llm_instance = TxtaiLLM(
    "LiteLLMs/gemma-2b-it-GGUF/Q8_0/Q8_0-00001-of-00001.gguf"
)

# =========================
# LANGCHAIN WRAPPER
# =========================
class TxtaiLangChainLLM(LangChainLLM):

    @property
    def _llm_type(self) -> str:
        return "txtai"

    def _generate(
        self,
        prompts: List[str],
        stop: Optional[List[str]] = None,
        **kwargs: Any
    ) -> LLMResult:

        generations = []

        for prompt in prompts:
            response = txtai_llm_instance(prompt)
            generations.append(
                [Generation(text=response)]
            )

        return LLMResult(generations=generations)


llm = TxtaiLangChainLLM()

# =========================
# MOOD BOARD OPTIONS
# =========================
MOOD_BOARD_OPTIONS = [
    "Software Engineer", "Data Scientist", "AI Enthusiast", "Entrepreneur",
    "Designer", "Gaming", "Reading", "Gym/Fitness", "Traveling",
    "Photography", "Anime", "Movies", "Netflix", "Music", "Coffee",
    "Tea", "Pizza", "Street Food", "Chocolate", "Dark Humor",
    "Sarcasm", "Wholesome Memes", "Gen-Z Humor", "Nerd Humor"
]

# =========================
# MEME DATING PROMPT
# =========================
MEME_DATING_PROMPT = """
You are the Ultimate Meme Dating App AI Assistant.

Your mission is to determine whether two people are destined to become:
* Meme Soulmates
* Chaos Partners
* Co-op Teammates
* Main Character Duo
* Or a Complete Internet Disaster

Analyze both users using their interests, mood boards, and overall internet energy.

=================================================
MEME PERSONALITY ANALYSIS
=========================

Determine a Meme Personality Type for each user.

Use these archetypes when applicable:

Tech Interests
(AI, Coding, Programming, ML, Data Science, Startups)
→ The Caffeinated Tech Goblin

Gaming Interests
(Gaming, Esports, Streaming)
→ The Respawn Romantic

Fitness Interests
(Gym, Fitness, Sports, Workout)
→ The Pre-Workout Philosopher

Entertainment Interests
(Movies, Anime, Netflix, Series)
→ The Pop-Culture Archivist

Creative Interests
(Photography, Design, Art, Music)
→ The Creative Chaos Engineer

Books & Writing Interests
(Books, Reading, Poetry, Writing)
→ The Overthinking Storyteller

Travel & Adventure Interests
(Traveling, Hiking, Adventure)
→ The Wandering Memory Collector

Food & Beverage Interests
(Coffee, Tea, Pizza, Street Food, Chocolate)
→ The Comfort-Core Connoisseur

If multiple categories overlap:
Create a unique hybrid archetype.

For each user infer:
* Personality Traits
* Humor Style
* Communication Style
* Internet Behavior
* Meme Personality Type

=================================================
COMPATIBILITY SCORING
=====================

Generate a compatibility score between 0 and 100.

Scoring Rules:
Shared Interests > 4 -> Score between 85-100
Shared Interests between 2 and 4 -> Score between 70-85
Shared Interests = 1 -> Score between 55-70
No common interests but complementary personalities -> Score between 50-75
No common interests and conflicting personalities -> Score below 50

=================================================
IMPORTANT
=========
Return ONLY the requested output.
Do NOT explain your reasoning.
Do NOT repeat rules.
Do NOT mention scoring logic.
Do NOT add extra sections.

=================================================
OUTPUT FORMAT
=============
Follow this exact formatting. Replace names dynamically with the names provided in the input.

## {user_1_name} and {user_2_name}'s Meme Match Report

**MEME PERSONALITY:**

* User 1: <1-sentence description of User 1's vibe, traits, and archetype>
* User 2: <1-sentence description of User 2's vibe, traits, and archetype>

**COMPATIBILITY SCORE:** <0-100>%

**WHY THEY MATCH:**

<2-3 concise sentences explaining how their specific interests or personalities mesh together dynamically.>

**SHARED INTERESTS:**

<Bullet points of common interests, or if none, bullet points of complementary interest paths>

**MEME PICKUP LINE:**

"<A short, funny, copy-paste ready meme pickup line using their interests>"

**ICEBREAKER:**

"<A witty, natural opening message utilizing their interests with smooth emoji use>"

=================================================
INPUT
=====

User 1 Name:
{user_1_name}

User 1 Age:
{user_1_age}

User 1 Gender:
{user_1_gender}

User 1 Interests:
{user_1_interests}

---

User 2 Name:
{user_2_name}

User 2 Age:
{user_2_age}

User 2 Gender:
{user_2_gender}

User 2 Interests:
{user_2_interests}

"""

# =========================
# GRADIO INTERFACE LOGIC
# =========================
def run_matchmaking_ui(
    u1_name, u1_age, u1_gender, u1_interests, u1_custom,
    u2_name, u2_age, u2_gender, u2_interests, u2_custom
):
    # Process User 1 Interests
    final_u1_interests = list(u1_interests) if u1_interests else []
    if u1_custom.strip():
        final_u1_interests.extend([i.strip() for i in u1_custom.split(",") if i.strip()])

    # Process User 2 Interests
    final_u2_interests = list(u2_interests) if u2_interests else []
    if u2_custom.strip():
        final_u2_interests.extend([i.strip() for i in u2_custom.split(",") if i.strip()])

    # Run Matchmaking Chain
    prompt = PromptTemplate.from_template(MEME_DATING_PROMPT)
    chain = prompt | llm | StrOutputParser()

    report = chain.invoke({
        "user_1_name": u1_name,
        "user_1_age": str(u1_age),
        "user_1_gender": u1_gender,
        "user_1_interests": ", ".join(final_u1_interests),

        "user_2_name": u2_name,
        "user_2_age": str(u2_age),
        "user_2_gender": u2_gender,
        "user_2_interests": ", ".join(final_u2_interests),
    })

    header = f"👤 {u1_name}  ❤️  {u2_name}"
    return header, report.strip()


# =========================
# GRADIO LAYOUT
# =========================
with gr.Blocks(theme=gr.themes.Soft()) as app:
    gr.Markdown("# 💘 LLMates AI : Where Matching Internet Souls, One Meme at a Time")

    with gr.Row():
        # LEFT SIDE INPUTS
        with gr.Column(scale=1):
            gr.Markdown("### 📊 Profiles Entry Setup")

            with gr.Tab("User 1 Details"):
                u1_name = gr.Textbox(label="Name", placeholder="Enter name...")
                u1_age = gr.Number(label="Age", value=None, precision=0)
                u1_gender = gr.Textbox(label="Gender", placeholder="e.g. Male, Female, Non-binary")
                u1_interests = gr.Dropdown(
                    label="Mood Board Choices",
                    choices=MOOD_BOARD_OPTIONS,
                    multiselect=True,
                    info="Select standard interests"
                )
                u1_custom = gr.Textbox(label="✨ Custom Interests", placeholder="Comma separated, e.g. Politics, Tech")

            with gr.Tab("User 2 Details"):
                u2_name = gr.Textbox(label="Name", placeholder="Enter name...")
                u2_age = gr.Number(label="Age", value=None, precision=0)
                u2_gender = gr.Textbox(label="Gender", placeholder="e.g. Male, Female, Non-binary")
                u2_interests = gr.Dropdown(
                    label="Mood Board Choices",
                    choices=MOOD_BOARD_OPTIONS,
                    multiselect=True,
                    info="Select standard interests"
                )
                u2_custom = gr.Textbox(label="✨ Custom Interests", placeholder="Comma separated")

            btn = gr.Button("🚀 Match Profiles", variant="primary")

        # RIGHT SIDE OUTPUTS
        with gr.Column(scale=2):
            gr.Markdown("### 💘 MATCH RESULT")

            match_header = gr.Textbox(
                label="Match Partners",
                interactive=False,
                show_copy_button=True
            )

            report_output = gr.Markdown(value="*Results will appear here after calculation.*")

    # Wire up function execution
    btn.click(
        fn=run_matchmaking_ui,
        inputs=[
            u1_name, u1_age, u1_gender, u1_interests, u1_custom,
            u2_name, u2_age, u2_gender, u2_interests, u2_custom
        ],
        outputs=[match_header, report_output]
    )

# =========================
# ENTRY POINT
# =========================
if __name__ == "__main__":
    app.launch()
```
